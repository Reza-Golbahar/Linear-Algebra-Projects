## A Linear Algebra Lab on the Power of Eigenvectors

This notebook links two seemingly unrelated applications through **one and the
same piece of linear algebra**: the dominant eigenvector of a matrix, computed
by **power iteration**.


## 1. The Shared Engine: Power Iteration with Deflation

Given a square matrix $A \in \mathbb{R}^{m\times m}$, power iteration finds the
eigenvector associated with the **largest-magnitude eigenvalue** by repeatedly
applying $A$ to a random vector and renormalizing:

$$ v_{t+1} = \frac{A v_t}{\lVert A v_t \rVert_2} $$

To find the **top-$k$** eigenpairs of a symmetric matrix, we use **deflation**:
once we have found eigenpair $(\lambda_1, v_1)$, we form

$$ A^{(1)} = A - \lambda_1 v_1 v_1^T $$

which has the same eigenvectors as $A$ except that the component along $v_1$
has been removed (its eigenvalue is now $0$). Repeating this on $A^{(1)}$
yields $(\lambda_2, v_2)$, and so on.

This trick **only works because $A$ is symmetric** (its eigenvectors are
orthogonal). Keep this in mind — you will be asked later why PageRank's matrix
cannot be deflated the same way.


```python

def power_iteration(A, num_iters=1000, tol=1e-10, v0=None, seed=None):
    """
    Compute the dominant eigenpair (largest |eigenvalue|) of square matrix A.

    Parameters
    ----------
    A : (m, m) ndarray
    num_iters : maximum number of iterations
    tol : stop when ||v_{t+1} - v_t|| < tol
    v0 : optional initial vector
    seed : optional random seed for the default initial vector

    Returns
    -------
    eigenvalue : float, the Rayleigh-quotient estimate of the dominant eigenvalue
    eigenvector : (m,) ndarray, unit-norm
    n_iter : int, number of iterations actually performed
    history : list of floats, ||v_{t+1}-v_t|| at every iteration (for convergence plots)
    """
    # TODO
    v0_dimension = A.shape[0]

    if v0 is None:
        if seed is None:
            random_number_generator = np.random.default_rng()
        else:
            random_number_generator = np.random.default_rng(seed)

        v0 = random_number_generator.random(v0_dimension)

    # Power iteration assumes the vector has unit length.
    # So, we must normalize our initial vector.
    v0_normalized = v0 / np.linalg.norm(v0)

    eigenvector = v0_normalized
    n_iter = 0
    history = []

    while n_iter < num_iters:
        w = A @ eigenvector
        eigenvector_next = w / np.linalg.norm(w)

        # Fix the arbitrary sign of the eigenvector
        if np.dot(eigenvector_next, eigenvector) < 0:
            eigenvector_next = -eigenvector_next

        n_iter += 1

        difference = eigenvector_next - eigenvector
        difference_norm = np.linalg.norm(difference)
        history.append(difference_norm)

        if difference_norm < tol:
            eigenvector = eigenvector_next
            break

        eigenvector = eigenvector_next

    # Rayleigh-quotient estimate
    eigenvalue = eigenvector @ (A @ eigenvector)

    return eigenvalue, eigenvector, n_iter, history


def top_k_eigenpairs_symmetric(A, k, num_iters=2000, tol=1e-10, seed=0):
    """
    Compute the top-k eigenpairs of a SYMMETRIC matrix A using power
    iteration + deflation. Eigenvalues are returned in decreasing order
    of magnitude.
    """
    # TODO
    A_copy = A.copy()

    eigvals = []
    eigvecs = []

    for i in range(k):
        eigenvalue, eigenvector, n_iter, history = power_iteration(
            A_copy,
            num_iters=num_iters,
            tol=tol,
            seed=seed + i
        )

        eigvals.append(eigenvalue)
        eigvecs.append(eigenvector)

        # Deflation
        A_copy = A_copy - eigenvalue * np.outer(
            eigenvector, eigenvector
        )

    eigvals = np.array(eigvals)
    eigvecs = np.column_stack(eigvecs)

    return eigvals, eigvecs
```

```python

# --- Sanity check against numpy's exact eigendecomposition ---
np.random.seed(1)
B = np.random.randn(6, 6)
B = B + B.T  # make it symmetric

vals_np, vecs_np = np.linalg.eigh(B)
order = np.argsort(-np.abs(vals_np))
vals_np_sorted = vals_np[order]

vals_pi, vecs_pi = top_k_eigenpairs_symmetric(B, k=3, seed=42)

print("numpy eigh (top 3 |eigenvalue|): ", np.round(vals_np_sorted[:3], 6))
print("power iteration (top 3):         ", np.round(vals_pi, 6))

match_3 = np.allclose(np.sort(np.abs(vals_np_sorted[:3])), np.sort(np.abs(vals_pi)), atol=1e-4)
print("Power iteration matches numpy:", match_3)
```

```python
# --- Sanity check against numpy's exact eigendecomposition ---
np.random.seed(99)
B2 = np.random.randn(10, 10)
B2 = B2 + B2.T

vals_np, vecs_np = np.linalg.eigh(B2)
order = np.argsort(-np.abs(vals_np))
vals_np = vals_np[order][:5]
vecs_np = vecs_np[:, order][:, :5]


vals_pi, vecs_pi = top_k_eigenpairs_symmetric(B2, k=5, num_iters=4000, seed=7)

match_vals = np.allclose(vals_np, vals_pi, atol=1e-3)
print("10x10 case, top-5 |eigenvalues| (numpy): ", np.round(vals_np, 5))
print("10x10 case, top-5 |eigenvalues| (power) :", np.round(vals_pi, 5))
print("Eigenvalues Match:", match_vals)

match_vecs = np.allclose(np.abs(vecs_np), np.abs(vecs_pi), atol=1e-3)
print("Eigenvectors Match:", match_vecs)

gb.check("top_k_eigenvalues_symmetric matches numpy (10x10, top-5)", match_vals, 5)
gb.check("top_k_eigenvectors_symmetric matches numpy (10x10, top-5)", match_vecs, 5)
```


### 1.1 Visualizing convergence of a single power iteration run

Before moving to real data, let's look at *how* power iteration converges:
at every step we plot $\lVert v_{t+1}-v_t\rVert$, and also the running estimate
of the eigenvalue (the Rayleigh quotient), to see both settle down.


```python

np.random.seed(3)
C_demo = np.random.randn(20, 20)
C_demo = C_demo + C_demo.T

lam_demo, v_demo, n_iter_demo, hist_demo = power_iteration(C_demo, num_iters=200, tol=1e-12, seed=5)

# Track the Rayleigh-quotient estimate at every iteration too
rq_trace = []
v_trace = v_demo.copy()
v_t = np.random.default_rng(5).standard_normal(20)
v_t = v_t / np.linalg.norm(v_t)
for _ in range(n_iter_demo):
    w = C_demo @ v_t
    w = w / np.linalg.norm(w)
    if np.dot(w, v_t) < 0:
        w = -w
    v_t = w
    rq_trace.append(v_t @ C_demo @ v_t)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].semilogy(hist_demo, marker=".")
axes[0].set_xlabel("Iteration")
axes[0].set_ylabel(r"$\|v_{t+1}-v_t\|_2$ (log scale)")
axes[0].set_title("Convergence of the eigenvector estimate")
axes[0].grid(alpha=0.3)

axes[1].plot(rq_trace, marker=".", color="darkorange")
axes[1].axhline(lam_demo, color="black", linestyle="--", label=f"final estimate = {lam_demo:.4f}")
axes[1].set_xlabel("Iteration")
axes[1].set_ylabel("Rayleigh quotient (eigenvalue estimate)")
axes[1].set_title("Convergence of the eigenvalue estimate")
axes[1].legend()
axes[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
```


## 2. Application 1 — Eigenfaces for Face Recognition

We use the **Olivetti faces** dataset: 400 grayscale images (64×64 pixels),
40 distinct people, 10 images per person.

The pipeline:

1. Flatten each image into a vector in $\mathbb{R}^{4096}$.
2. Split into train/test sets.
3. **Mean-center** the training images.
4. Compute the eigenfaces — the eigenvectors of the **covariance matrix** of
   the centered training data — using `top_k_eigenpairs_symmetric`.
5. Project every face (train and test) onto the top-$k$ eigenfaces. This is
   dimensionality reduction: each $4096$-dimensional image becomes a
   $k$-dimensional coefficient vector.
6. Classify a test face by **nearest neighbor** in this $k$-dimensional
   coefficient space.


```python

faces = fetch_olivetti_faces(shuffle=True, random_state=0)
X_all = faces.data            # shape (400, 4096), pixel values in [0, 1]
y_all = faces.target          # shape (400,), person id in [0, 39]
img_h, img_w = faces.images.shape[1], faces.images.shape[2]

print("Data matrix shape:", X_all.shape)
print("Number of distinct people:", len(np.unique(y_all)))

X_train, X_test, y_train, y_test = train_test_split(
    X_all, y_all, test_size=0.25, random_state=0, stratify=y_all
)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

```python

def center_data(X):
    """Return (X_centered, mean_vector). X is (n_samples, n_features)."""
    # TODO
    # Make a copy so that the original X is not modified.
    X_copy = X.copy()

    # Computing mean vector
    sum_vector = X[0].copy()

    for row_number in range(1, X.shape[0]):
        sum_vector += X[row_number]

    mean_vector = sum_vector / X.shape[0]

    # Subtract mean from every row
    for row_number in range(X.shape[0]):
        X_copy[row_number] = X[row_number] - mean_vector

    return X_copy, mean_vector


def covariance_small(Xc):
    """
    Compute the (n_samples x n_samples) Gram matrix L = Xc Xc^T / n.

    We deliberately do NOT form the (n_features x n_features) covariance
    matrix C = Xc^T Xc / n directly: with n_features = 4096 that matrix is
    4096x4096 (huge), while n_samples = 300 (small). L and C share the same
    nonzero eigenvalues, and their eigenvectors are related by a simple
    linear map (see the discussion question below) -- this is the classical
    "Turk & Pentland" trick that makes eigenfaces computationally feasible.
    """
    # TODO
    return (Xc @ np.transpose(Xc)) / Xc.shape[0]


def eigenfaces_from_small_eigvecs(Xc, eigvecs_small, eigvals_small):
    """
    Convert eigenvectors u_i of the small Gram matrix L = Xc Xc^T / n
    into eigenvectors of the true (huge) covariance matrix C = Xc^T Xc / n,
    via v_i = Xc^T u_i, renormalized to unit length.
    """
    # TODO
    eigenfaces = np.transpose(Xc) @ eigvecs_small

    # Normalize columns
    for column_number in range(eigenfaces.shape[1]):
        eigenfaces[:, column_number] = (
            eigenfaces[:, column_number]
            / np.linalg.norm(eigenfaces[:, column_number])
        )

    return eigenfaces
```

```python

K = 60  # number of eigenfaces to keep

Xc_train, mean_face = center_data(X_train)
L = covariance_small(Xc_train)

eigvals_small, eigvecs_small = top_k_eigenpairs_symmetric(L, k=K, num_iters=3000, tol=1e-9, seed=0)
eigenfaces = eigenfaces_from_small_eigvecs(Xc_train, eigvecs_small, eigvals_small)

print("Eigenface matrix shape:", eigenfaces.shape)   # (4096, K)
print("Top 10 eigenvalues:", np.round(eigvals_small[:10], 4))

# Eigenvalues of a covariance matrix must be non-negative (it is PSD); a
# negative entry here would indicate a numerical bug.
all_nonneg = np.all(eigvals_small > -1e-6)
gb.check("Eigenvalues of the covariance matrix are non-negative (PSD check)", all_nonneg, 3);
```


### 2.1 Scree plot and explained variance

The eigenvalues of the covariance matrix tell us how much variance each
eigenface direction explains. Plotting them in decreasing order (a **scree
plot**) and their cumulative sum tells us how many eigenfaces we *actually*
need to capture, say, 85% of the total variance in the training faces.


```python

total_variance = X_train.var(axis=0).sum()
cumulative_variance_ratio = np.cumsum(eigvals_small) / total_variance

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(np.arange(1, K + 1), eigvals_small, marker="o", markersize=3)
axes[0].set_xlabel("Eigenface index")
axes[0].set_ylabel("Eigenvalue (variance explained)")
axes[0].set_title("Scree plot")
axes[0].grid(alpha=0.3)

axes[1].plot(np.arange(1, K + 1), cumulative_variance_ratio, marker="o", markersize=3, color="seagreen")
axes[1].axhline(0.85, color="red", linestyle="--", label="85% threshold")
axes[1].set_xlabel("Number of eigenfaces K")
axes[1].set_ylabel("Cumulative fraction of variance explained")
axes[1].set_title("Cumulative explained variance")
axes[1].legend()
axes[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()

K_for_85 = int(np.searchsorted(cumulative_variance_ratio, 0.85) + 1)
print(f"Number of eigenfaces needed to explain 85% of training variance: {K_for_85}"
      f" (out of the {K} we computed; cumulative variance at K={K} is "
      f"{cumulative_variance_ratio[-1]*100:.1f}%)")
```

```python

def project(X, mean_face, eigenfaces):
    """Project rows of X onto the eigenface basis. Returns (n_samples, K) coefficients."""
    return (X - mean_face) @ eigenfaces


def reconstruct(coeffs, mean_face, eigenfaces):
    """Reconstruct images (n_samples, n_features) from their coefficients."""
    return coeffs @ eigenfaces.T + mean_face


train_coeffs = project(X_train, mean_face, eigenfaces)
test_coeffs = project(X_test, mean_face, eigenfaces)

reconstructed = reconstruct(train_coeffs[:8], mean_face, eigenfaces)

fig, axes = plt.subplots(2, 8, figsize=(12, 3.2))
for i in range(8):
    axes[0, i].imshow(X_train[i].reshape(img_h, img_w), cmap="gray")
    axes[0, i].axis("off")
    axes[1, i].imshow(reconstructed[i].reshape(img_h, img_w), cmap="gray")
    axes[1, i].axis("off")
plt.suptitle(f"Original (top row) vs. reconstruction (bottom row) using K={K} eigenfaces")
plt.tight_layout()
plt.show()
```


### 2.2 Reconstruction error as a function of K

Eckart–Young guarantees that more eigenfaces means a (provably) better
reconstruction on the *training* data. Let's measure this directly with the
mean-squared reconstruction error, instead of just eyeballing images.


```python

ks_recon = [1, 2, 4, 8, 16, 32, 60]
mse_train, mse_test = [], []
for k in ks_recon:
    ef_k = eigenfaces[:, :k]
    tr_c = project(X_train, mean_face, ef_k)
    te_c = project(X_test, mean_face, ef_k)
    tr_rec = reconstruct(tr_c, mean_face, ef_k)
    te_rec = reconstruct(te_c, mean_face, ef_k)
    mse_train.append(np.mean((X_train - tr_rec) ** 2))
    mse_test.append(np.mean((X_test - te_rec) ** 2))

plt.figure(figsize=(6.5, 4))
plt.plot(ks_recon, mse_train, marker="o", label="train MSE")
plt.plot(ks_recon, mse_test, marker="s", label="test MSE")
plt.xlabel("Number of eigenfaces K")
plt.ylabel("Mean squared reconstruction error")
plt.title("Reconstruction error vs. number of components")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Train MSE must be monotonically non-increasing in K -- this is exactly what
# Eckart-Young predicts (adding components can only help the training fit).
train_mse_monotonic = all(mse_train[i] >= mse_train[i+1] - 1e-8 for i in range(len(mse_train)-1))
gb.check("Training reconstruction MSE is non-increasing in K (Eckart-Young)", train_mse_monotonic, 4);
```

```python

def nearest_neighbor_classify(train_coeffs, train_labels, test_coeffs):
    """
    Classify every row of test_coeffs by 1-nearest-neighbor (Euclidean
    distance) against train_coeffs. Pure linear algebra: no sklearn classifier.
    """
    # TODO
    y_pred = []

    for test_index in range(test_coeffs.shape[0]):
        error = 100_000
        train_index = -1

        test_data = test_coeffs[test_index]

        for new_train_index in range(train_coeffs.shape[0]):
            new_error = np.linalg.norm(
                test_data - train_coeffs[new_train_index]
            )

            if new_error < error:
                error = new_error
                train_index = new_train_index

        y_pred.append(train_labels[train_index])

    return y_pred


y_pred = nearest_neighbor_classify(train_coeffs, y_train, test_coeffs)
acc = accuracy_score(y_test, y_pred)
print(f"Eigenfaces + 1-NN test accuracy with K={K}: {acc * 100:.2f}%")

gb.check(f"Eigenfaces + 1-NN test accuracy >= 80% (K={K})", acc >= 0.80, 10, detail=f"(got {acc*100:.1f}%)")
gb.check(f"Eigenfaces + 1-NN test accuracy >= 90% (K={K})", acc >= 0.90, 5, detail=f"(got {acc*100:.1f}%, bonus)");
```


### 2.5 Accuracy vs. number of eigenfaces, on a finer grid of K

Now let's repeat the K-sweep from before, but record the actual accuracy
numbers explicitly (rather than just plotting), and view the curve on both a

### Answers Here:

1. Let

$$
X_c \in \mathbb{R}^{n\times d},
$$

where (n) is the number of training images and (d) is the number of pixels, with (d \gg n). Define

$$
L=\frac{1}{n}X_cX_c^T,
\qquad
C=\frac{1}{n}X_c^TX_c.
$$

Assume that (u) is a unit eigenvector of (L) with eigenvalue (\lambda\neq0):

$$
Lu=\lambda u.
$$

Substituting the definition of (L),

$$
\frac{1}{n}X_cX_c^Tu=\lambda u.
$$

Multiplying both sides by (X_c^T),

$$
\frac{1}{n}X_c^TX_c(X_c^Tu)
=

\lambda(X_c^Tu).
$$

Since

$$
C=\frac{1}{n}X_c^TX_c,
$$

we obtain

$$
C(X_c^Tu)=\lambda(X_c^Tu).
$$

Therefore,

$$
v=X_c^Tu
$$

is an eigenvector of the covariance matrix (C) corresponding to the same eigenvalue (\lambda). Since (v) is generally not a unit vector, it is normalized before being used as an eigenface:

$$
v_{\text{normalized}}
=

\frac{X_c^Tu}{\left|X_c^Tu\right|_2}.
$$

The covariance matrix (C) is (d\times d), but its rank satisfies

$$
\operatorname{rank}(C)
=

\operatorname{rank}(X_c)
\leq n.
$$

Consequently, (C) has at most (n) nonzero eigenvalues. All remaining eigenvalues are zero, and their eigenvectors span directions in which the training data has no variance.

These directions contain no information about variation in the training images and therefore contribute nothing to PCA. Since (d>n), computing the much smaller matrix (L) allows us to recover every nonzero eigenvalue and its corresponding informative principal component without constructing the enormous (d\times d) covariance matrix.

2. The Eckart–Young theorem states that if the singular value decomposition of the centered data matrix is

$$
X_c=U\Sigma V^T,
$$

then the best rank-(K) approximation of (X_c), under either the Frobenius norm or the spectral norm, is obtained by keeping only the largest (K) singular values:

$$
X_{c,K}
=

U_K\Sigma_KV_K^T.
$$

More precisely,

$$
X_{c,K}
=

\underset{\operatorname{rank}(Y)\leq K}{\arg\min}
\left|X_c-Y\right|_F.
$$

The eigenvalues of the covariance matrix are related to the singular values by

$$
\lambda_i=\frac{\sigma_i^2}{n},
$$

where (\sigma_i) is the (i)-th singular value of (X_c).

Therefore, selecting the top (K) eigenfaces is equivalent to retaining the (K) largest singular values. The resulting approximation minimizes the total squared reconstruction error:

$$
\left|X_c-X_{c,K}\right|_F^2
=

\sum_{i=K+1}^{r}\sigma_i^2,
$$

where

$$
r=\operatorname{rank}(X_c).
$$

Hence, the retained eigenfaces provide the **provably optimal rank-(K) linear reconstruction of the training data** under the Frobenius norm.

However, this guarantee applies only to the training set used to compute the eigenfaces. The theorem does not guarantee optimal reconstruction of unseen test images because the principal components were estimated only from the training data.

Therefore, the training reconstruction error must be monotonically non-increasing as (K) increases, while the test reconstruction error does not have the same theoretical guarantee. The test error may decrease more slowly, level off, or behave differently because the test images are unseen data.

3. Let

$$
\mu=\frac{1}{n}\sum_{i=1}^{n}x_i
$$

be the mean face, and let

$$
X_c=X-\mathbf{1}\mu^T,
$$

where (\mathbf{1}) is the (n)-dimensional vector whose entries are all (1).

Therefore,

$$
X=X_c+\mathbf{1}\mu^T.
$$

The uncentered covariance-like matrix is

$$
X^TX.
$$

Substituting the expression for (X),

$$
X^TX
=

\left(X_c+\mathbf{1}\mu^T\right)^T
\left(X_c+\mathbf{1}\mu^T\right).
$$

Expanding the product,

$$
X^TX
=

X_c^TX_c
+
X_c^T\mathbf{1}\mu^T
+
\mu\mathbf{1}^TX_c
+
\mu\mathbf{1}^T\mathbf{1}\mu^T.
$$

Because the data is centered,

$$
X_c^T\mathbf{1}=0.
$$

Therefore, the two cross terms vanish. Also,

$$
\mathbf{1}^T\mathbf{1}=n.
$$

Thus,

$$
X^TX
=

X_c^TX_c+n\mu\mu^T.
$$

Therefore, skipping centering adds the rank-1 term

$$
n\mu\mu^T.
$$

If the mean face has a much larger norm than the variations between individuals, this additional term can dominate the covariance-like matrix.

As a result, the leading principal component tends to point toward the mean face rather than toward directions representing differences between individuals. The first eigenface would therefore capture common facial structure and overall brightness rather than identity-related variation, reducing the effectiveness of PCA for face recognition.

4. The current classifier uses Euclidean distance between two coefficient vectors (a) and (b):

$$
d_E(a,b)
=

\sqrt{
\sum_{i=1}^{K}(a_i-b_i)^2
}.
$$

This treats every eigenface direction equally.

A Mahalanobis-style weighted distance instead scales each coordinate according to the inverse of its standard deviation:

$$
d_M(a,b)
=

\sqrt{
\sum_{i=1}^{K}
\frac{(a_i-b_i)^2}{\lambda_i}
}.
$$

Equivalently, if

$$
\Lambda=
\operatorname{diag}
\left(
\lambda_1,\lambda_2,\ldots,\lambda_K
\right),
$$

then

$$
d_M(a,b)
=

\left|
\Lambda^{-1/2}(a-b)
\right|_2.
$$

Geometrically, PCA transforms the data into axes aligned with the covariance ellipsoid.

Euclidean distance measures displacement equally in every principal direction. In contrast, Mahalanobis distance gives less weight to directions with large variance and more weight to directions with small variance.

Thus, a difference along a high-variance eigenface direction contributes relatively little:

$$
\frac{(a_i-b_i)^2}{\lambda_i}
\quad\text{is small when }\lambda_i\text{ is large}.
$$

Conversely, a difference along a low-variance direction receives more weight.

This could improve recognition if the high-variance directions primarily represent nuisance factors such as illumination or pose. However, it could also reduce accuracy if those high-variance directions contain useful identity information.

Therefore, changing from Euclidean to Mahalanobis-style distance could change the accuracy-versus-(K) curve in either direction, depending on the structure of the dataset.

5. Consider two different individuals photographed under identical lighting conditions, while another photograph of the first individual is taken under strong side lighting.

The illumination change may produce much larger pixel differences than the actual differences between the two people's facial identities. Since PCA captures directions of maximum variance, many principal components represent lighting variation rather than identity. Consequently, the coefficients of the two differently illuminated images of the same person may differ more than the coefficients of two different people photographed under similar lighting.

This demonstrates that directions of largest variance are not necessarily the directions that best discriminate between classes. PCA is an unsupervised technique that ignores class labels and preserves overall variance, regardless of whether that variance is useful for recognition.

Linear Discriminant Analysis (LDA), used in the Fisherfaces method, addresses this limitation by maximizing the separation between different people while simultaneously minimizing variation within the same person's images. Thus, Fisherfaces explicitly optimize discrimination rather than reconstruction.

6. Suppose one individual is frequently misclassified as another in the confusion matrix.

A plausible explanation is that their projected coefficient vectors lie close together in the (K)-dimensional eigenface space. This can occur if the two individuals share similar global facial characteristics, such as face shape, eye spacing, hairstyle, or facial proportions. Since PCA retains directions of greatest overall variance rather than the directions that best separate identities, the projection may discard subtle but discriminative features.

Geometrically, each face is represented as a point in the (K)-dimensional coefficient space. The 1-nearest-neighbor classifier assigns a test image to the label of the closest training point according to Euclidean distance. If two individuals occupy overlapping or nearby regions of this space, a test image from one person may fall closer to the training samples of the other person than to samples of its true class, resulting in misclassification. This indicates that the PCA subspace preserves reconstruction quality but does not necessarily maximize class separability.


## 3. Application 2 — PageRank as an Eigenvector Problem

We model a small set of web pages as a directed graph: an edge $i \to j$
means "page $i$ links to page $j$". PageRank assigns every page a score
equal to the stationary probability that a "random surfer" — who, with
probability $d$ (the damping factor) follows a random outgoing link, and with
probability $1-d$ jumps to a uniformly random page — is on that page after
infinitely many steps.

This stationary distribution is the eigenvector of eigenvalue $1$ of the
**Google matrix**

$$ M = d\,S + \frac{1-d}{N}\mathbf{1}\mathbf{1}^T $$

where $S$ is the column-stochastic version of the link matrix (columns sum to
$1$; dangling nodes — pages with no outgoing links — are replaced by a
uniform column).

We reuse **the exact same `power_iteration` function** from Section 1.


```python

pages = ["A", "B", "C", "D", "E", "F"]
edges = [
    ("A", "B"), ("A", "C"),
    ("B", "C"),
    ("C", "A"),
    ("D", "C"), ("D", "B"),
    ("E", "D"), ("E", "F"),
    ("F", "E"),
]
pages.append("G")  # dangling node: G has zero outgoing links

N = len(pages)
idx = {p: i for i, p in enumerate(pages)}

G = nx.DiGraph()
G.add_nodes_from(pages)
G.add_edges_from(edges)

print(f"Graph has {G.number_of_nodes()} pages and {G.number_of_edges()} links.")
print("In-degrees: ", dict(G.in_degree()))
print("Out-degrees:", dict(G.out_degree()))
```

```python

def build_link_matrix(pages, edges):
    """
    Build the raw (unnormalized) link matrix H where H[j, i] = 1 if page i
    links to page j (note the transpose convention: columns = source page,
    rows = destination page, so that S @ rank_vector aggregates incoming rank).
    """
    # TODO
    dimension = len(pages)
    H = np.zeros((dimension, dimension))

    idx = {p: i for i, p in enumerate(pages)}

    for edge in edges:
        H[idx[edge[1]], idx[edge[0]]] = 1

    return H, idx


def make_column_stochastic(H):
    """
    Normalize each column to sum to 1. Columns that sum to 0 (dangling
    nodes, i.e. pages with no outgoing links) are replaced with a uniform
    distribution over all pages -- the standard fix so that S remains a
    valid column-stochastic (and hence Markov) matrix.
    """
    # TODO
    H = H.copy()  # IMPORTANT: don't modify the raw H

    dimension = H.shape[0]
    dangling = []

    for column_number in range(dimension):

        column_sum = 0

        for row_number in range(dimension):
            column_sum += H[row_number, column_number]

        if column_sum == 0:
            H[:, column_number] = np.ones(dimension) / dimension
            dangling.append(column_number)

        else:
            for row_number in range(dimension):
                H[row_number, column_number] = (
                    H[row_number, column_number] / column_sum
                )

    return H, dangling


def google_matrix(S, d=0.85):
    # TODO
    N = S.shape[0]
    
    M = d * S
    
    all_one_square_matrix = np.ones((N, N))
    M += ((1 - d) / N) * all_one_square_matrix
    
    return M
```

```python

H, idx = build_link_matrix(pages, edges)
S, dangling = make_column_stochastic(H)
print("Raw link matrix H:")
print(H.astype(int))
print("\nDangling nodes:", [pages[i] for i in dangling])

d = 0.85
M = google_matrix(S, d=d)

col_sums_S_ok = np.allclose(S.sum(axis=0), 1.0)
col_sums_M_ok = np.allclose(M.sum(axis=0), 1.0)
print("Column sums of S (should all be 1):", np.round(S.sum(axis=0), 6))
print("Column sums of M (should all be 1):", np.round(M.sum(axis=0), 6))

gb.check("S is column-stochastic (columns sum to 1)", col_sums_S_ok, 3)
gb.check("M is column-stochastic (columns sum to 1)", col_sums_M_ok, 3)
```

### 3.3 Visualizing convergence: how the rank vector evolves, iteration by iteration

```python

def power_iteration_trace(A, num_iters, v0=None, seed=None):
    """Like power_iteration, but returns the FULL sequence of vectors (for plotting)."""
    # TODO
    v0_dimension = A.shape[0]

    eigenvectors = []

    if v0 is None:
        if seed is None:
            random_number_generator = np.random.default_rng()
        else:
            random_number_generator = np.random.default_rng(seed)

        v0 = random_number_generator.random(v0_dimension)

    # Normalize initial vector
    v0_normalized = v0 / np.linalg.norm(v0)

    eigenvectors.append(v0_normalized)

    eigenvector = v0_normalized

    for i in range(num_iters):
        w = A @ eigenvector
        eigenvector_next = w / np.linalg.norm(w)

        # Fix the arbitrary sign of the eigenvector
        if np.dot(eigenvector_next, eigenvector) < 0:
            eigenvector_next = -eigenvector_next

        eigenvectors.append(eigenvector_next)
        eigenvector = eigenvector_next   

    return np.array(eigenvectors)


trace = power_iteration_trace(M, num_iters=40, seed=7)
trace = np.abs(trace)
trace = trace / trace.sum(axis=1, keepdims=True)  # normalize each row to a probability vector

plt.figure(figsize=(8, 5))
for i, p in enumerate(pages):
    plt.plot(trace[:, i], marker=".", label=p)
plt.xlabel("Iteration")
plt.ylabel("Estimated PageRank score")
plt.title("Evolution of each page's rank estimate over power-iteration steps")
plt.legend(ncol=4, fontsize=9)
plt.grid(alpha=0.3)
plt.show()
```

Power iteration's convergence rate is controlled by the ratio

### Answers Here:

1. The Perron–Frobenius theorem states that if a matrix is strictly positive, then it has a unique largest eigenvalue in magnitude. This eigenvalue is real and positive, and its corresponding eigenvector can be chosen to have all positive entries.

The Google matrix is

$$
M = dS + \frac{1-d}{N}\mathbf{1}\mathbf{1}^T,
$$

where \(0 < d < 1\). Since \(S\) is a non-negative matrix, every entry of \(dS\) is non-negative. Also, every entry of

$$
\frac{1-d}{N}\mathbf{1}\mathbf{1}^T
$$

is equal to

$$
\frac{1-d}{N} > 0.
$$

Therefore, for every \(i,j\),

$$
M_{ij}
=
dS_{ij} + \frac{1-d}{N}
>
0.
$$

Thus, \(M\) is strictly positive.

Since \(M\) is strictly positive, the Perron–Frobenius theorem guarantees that its dominant eigenvalue is unique and has a strictly positive eigenvector. We also know that \(M\) is column-stochastic because

$$
\sum_{i=1}^{N} M_{ij} = 1
$$

for every column \(j\). Therefore,

$$
\mathbf{1}^T M = \mathbf{1}^T.
$$

Hence \(1\) is an eigenvalue of \(M\). Since \(M\) is non-negative and column-stochastic, all of its eigenvalues satisfy

$$
|\lambda_i| \leq 1.
$$

Therefore, the Perron eigenvalue must be exactly

$$
\lambda_1 = 1.
$$

Consequently, there is a unique strictly positive vector \(r\), up to scaling, satisfying

$$
Mr = r.
$$

After normalizing its entries to sum to \(1\),

$$
\sum_{i=1}^{N} r_i = 1,
$$

the vector \(r\) is the PageRank vector.

At \(d=1\), the teleportation term disappears:

$$
M = S.
$$

The matrix \(S\) is only guaranteed to be non-negative and column-stochastic; it is no longer guaranteed to be strictly positive. Therefore, the Perron–Frobenius argument for uniqueness no longer applies. In particular, \(S\) can be reducible or periodic, so eigenvalue \(1\) may have multiplicity greater than one and power iteration may fail to converge to a unique ranking.

2. For the symmetric covariance matrix, if

$$
Av_i=\lambda_i v_i
$$

and the eigenvectors are orthonormal, then we can use the deflation step

$$
A_{\text{new}}
=

A-\lambda_i v_iv_i^T.
$$

This works because, for a symmetric matrix, the left and right eigenvectors are the same:

$$
u_i=v_i.
$$

For a general non-symmetric matrix, however, the right eigenvectors satisfy

$$
Av_i=\lambda_i v_i,
$$

while the left eigenvectors satisfy

$$
u_i^TA=\lambda_i u_i^T.
$$

In general,

$$
u_i\neq v_i.
$$

Therefore, subtracting

$$
\lambda_i v_iv_i^T
$$

is not the correct deflation operation for a general non-symmetric matrix.

The appropriate bi-orthogonal deflation would instead involve both the right and left eigenvectors:

$$
A_{\text{new}}
=

A-\lambda_i
\frac{v_i u_i^T}{u_i^Tv_i}.
$$

The denominator ensures the appropriate normalization when the left and right eigenvectors are not unit-normalized in a compatible way.

Thus, the simple deflation method used for eigenfaces relies specifically on symmetry. For PageRank, we only need the dominant right eigenvector, so there is no reason to perform deflation at all.


3. Power iteration converges approximately at a rate determined by

$$
\left|\frac{\lambda_2}{\lambda_1}\right|.
$$

For the Google matrix,

$$
\lambda_1=1
$$

and the known bound is

$$
|\lambda_2|\leq d.
$$

Therefore,

$$
\left|\frac{\lambda_2}{\lambda_1}\right|
\leq d.
$$

Suppose the error behaves approximately as

$$
|v_t-v^*|
\leq C d^t,
$$

where (C) depends on the initial vector and the matrix.

To guarantee

$$
|v_t-v^*|<\varepsilon,
$$

we require

$$
Cd^t<\varepsilon.
$$

Taking logarithms,

$$
\ln(Cd^t)<\ln(\varepsilon).
$$

Therefore,

$$
\ln C+t\ln d<\ln\varepsilon.
$$

Since (0<d<1),

$$
\ln d<0.
$$

Thus,

$$
t>\frac{\ln(\varepsilon/C)}{\ln d}.
$$

Equivalently,

$$
\boxed{
t>
\frac{\ln(C/\varepsilon)}
{\ln(1/d)}
}
$$

and therefore a sufficient iteration count is approximately

$$
\boxed{
t_{\min}
=

\left\lceil
\frac{\ln(C/\varepsilon)}
{\ln(1/d)}
\right\rceil.
}
$$

If we ignore the constant (C) to focus on the dependence on (d), this becomes approximately

$$
\boxed{
t_{\min}
\approx
\frac{\ln(1/\varepsilon)}
{\ln(1/d)}.
}
$$

Now consider what happens as (d\to1). Since

$$
\ln(1/d)\to0,
$$

we obtain

$$
t_{\min}\to\infty.
$$

For example, if

$$
d=0.85,
$$

then

$$
\ln(1/d)\approx0.1625.
$$

But if

$$
d=0.99,

then

$$
\ln(1/d)\approx0.01005.
$$

The denominator is therefore roughly **16 times smaller** at (d=0.99), so many more iterations may be required for the same tolerance. This explains why the convergence plot becomes dramatically slower as (d) approaches (1).

Therefore, (C) can receive substantially more PageRank under the self-loop interpretation.

The uniform fix is more faithful to the standard **random-surfer interpretation** because a user who reaches a page with no outgoing links does not have a meaningful link to follow. ### Answer Here:

1. Eigenfaces and PageRank both use eigenvectors to extract important structure from a matrix, but they solve different problems. In eigenfaces, we work with the covariance matrix

$$
C = \frac{1}{n}X_c^T X_c,
$$

which is symmetric and positive semi-definite, so its eigenvalues are real and non-negative and its eigenvectors can be chosen to be orthogonal. We want the top \(K\) eigenvectors because they correspond to the \(K\) directions containing the largest amounts of variance in the training faces. Since the covariance matrix is symmetric, we can use deflation: after finding an eigenpair \((\lambda_i,v_i)\), we remove its contribution using

$$
A_{\text{new}} = A - \lambda_i v_i v_i^T.
$$

This works because, for a symmetric matrix, the left and right eigenvectors are the same and eigenvectors corresponding to different eigenvalues are orthogonal. In PageRank, the Google matrix \(M\) is generally not symmetric, and we only need one eigenvector: the dominant right eigenvector corresponding to eigenvalue \(1\). Therefore, we do not need deflation and simply use power iteration to find the dominant eigenvector. In ordinary power iteration, normalization is arbitrary because multiplying an eigenvector by a nonzero constant does not change its direction, whereas PageRank requires a probability distribution, so we normalize the final vector so that its entries sum to \(1\). The damping factor has a different role: for \(0<d<1\), it makes every entry of the Google matrix positive and guarantees a unique positive PageRank vector.

2. Computing a \(4096\times4096\) covariance matrix with a dense eigendecomposition is already expensive, but computing PageRank of a million-page web graph using `numpy.linalg.eig` on the full \(10^6\times10^6\) Google matrix would fail much more dramatically. A dense \(n\times n\) eigendecomposition requires approximately

$$
O(n^3)
$$

time and

$$
O(n^2)
$$

memory. A \(10^6\times10^6\) dense matrix contains

$$
10^{12}
$$

entries, requiring approximately

$$
10^{12}\times 8 = 8\times10^{12}\text{ bytes}
$$

or about \(8\) TB just to store the matrix using 64-bit floating-point numbers. A dense eigendecomposition would additionally require approximately

$$
O((10^6)^3)=O(10^{18})
$$

operations, making it completely impractical. The web graph is naturally sparse because each page usually links to only a small number of other pages, so we should not explicitly construct the full dense Google matrix. A power-iteration step only requires a matrix-vector multiplication, whose cost for a sparse matrix is approximately

$$
O(\operatorname{nnz}(M)),
$$

where \(\operatorname{nnz}(M)\) is the number of nonzero entries. If each page has only a small number of outgoing links, then

$$
\operatorname{nnz}(M)=O(N),
$$

so one power-iteration step can be approximately linear in the number of pages rather than quadratic. Similarly, for eigenfaces, instead of directly diagonalizing the \(4096\times4096\) covariance matrix, we use the \(X_cX_c^T\) trick to construct a much smaller \(300\times300\) matrix when there are only \(300\) training images. Therefore, the million-page PageRank problem would fail first and by an enormous margin if we attempted a dense eigendecomposition, while sparse power iteration makes the PageRank computation feasible.
